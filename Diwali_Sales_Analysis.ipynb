```python
import pandas as pd
df=pd.read_csv('Diwali Sales Data.csv',encoding='cp1252')
df
```

```python
df.info()
```

```python
df.shape
```

```python
df.isnull().sum() #to fins out total null values in each columns
```

```python
df=df.dropna(axis=1,how='all') # to  delete columns who is no having any value(using how='all')
df.columns.tolist()#to check remainig columns
```

```python
# Check missing Amount values
print(df['Amount'].isnull().sum())

# Remove rows where Amount is missing
df = df.dropna(subset=['Amount'])

# Verify
print(df['Amount'].isnull().sum())
print(df.shape)
```

```python
# Count duplicate rows
duplicate_count = df.duplicated().sum()

print("Total duplicate rows:", duplicate_count)
```

```python
df[df.duplicated()] # to check rows of duplicate values
```

```python
# Show all copies of duplicated rows
df[df.duplicated(keep=False)].sort_values(
    by=df.columns.tolist()
)
```

```python
# Remove exact duplicate rows
df = df.drop_duplicates()

# Check duplicate count again
print("Duplicates after cleaning:", df.duplicated().sum())

# Check dataset size
print("Dataset shape:", df.shape)
```

```python
df.info()
```

```python
df.groupby('Gender')['Amount'].sum()
```

```python
gender_sales = df.groupby('Gender')['Amount'].sum()

gender_sales.plot(kind='bar', title='Total Sales by Gender')
```

```python
age_sales = df.groupby('Age Group')['Amount'].sum()

age_sales
```

```python
age_sales.plot(
    kind='bar',
    title='Total Sales by Age Group',
    xlabel='Age Group',
    ylabel='Total Amount Spent',
    rot=45
)
```

```python
category_sales=df.groupby('Product_Category')['Amount'].sum().sort_values(ascending=False)
category_sales
```

```python
top_5 = category_sales.head(5)

top_5.plot(
    kind='bar',
    title='Top 5 Product Categories by Sales',
    xlabel='Product Category',
    ylabel='Total Sales Amount',
    rot=30
)
```

```python
state_sales = (
    df.groupby('State')['Amount']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
state_sales
```

```python
state_sales.plot(
    kind='bar',
    title='Top 10 States by Sales',
    xlabel='State',
    ylabel='Total Sales Amount',
    rot=45
)
```

```python
occupation_sales=df.groupby('Occupation')['Amount'].sum().sort_values(ascending=False)
occupation_sales
```

```python
occupation_analysis = df.groupby('Occupation').agg(
    Total_Sales=('Amount', 'sum'),
    Customer_Count=('User_ID', 'nunique'),
    Average_Spending=('Amount', 'mean')
)

occupation_analysis.sort_values(
    by='Average_Spending',
    ascending=False
)
```

```python
customer_spending = df.groupby('User_ID').agg(
    Customer_Total=('Amount', 'sum')
)

print("Total unique customers:", len(customer_spending))

print("Average spending per customer:",
      round(customer_spending['Customer_Total'].mean(), 2))

print("Highest customer spending:",
      round(customer_spending['Customer_Total'].max(), 2))
```

```python
customer_spending['Spending_Category'] = pd.cut(
    customer_spending['Customer_Total'],
    bins=[0, 10000, 30000, float('inf')],
    labels=['Low (up to 10k)', 'Medium (10k–30k)', 'High (30k+)'],
    include_lowest=True
)

category_counts = customer_spending['Spending_Category'].value_counts(
    sort=False
)

print(category_counts)
```

```python
category_counts.plot(
    kind='bar',
    title='Customer Spending Segmentation',
    xlabel='Spending Category',
    ylabel='Number of Customers',
    rot=15
)
```

# Diwali Sales Data Analysis

## Project Overview
This project analyzes Diwali sales data using Python and Pandas to understand customer purchasing behaviour and identify important sales trends.

## Objectives
- Clean the dataset by handling missing values and duplicate records.
- Analyze sales by gender, age group, product category, state, and occupation.
- Calculate customer-level spending and segment customers by spending range.
- Visualize important findings and develop data-driven business recommendations.

## Tools and Technologies
- Python
- Pandas
- Matplotlib
- Google Colab

## Dataset Preparation
- Removed columns that contained no data.
- Removed records with missing `Amount` values.
- Identified and removed duplicate rows.
- Prepared the cleaned dataset for exploratory data analysis (EDA).

## Key Findings

1. **Gender Analysis:** Female customers contributed approximately ₹7.43 crore in total sales, compared with approximately ₹3.19 crore from male customers.

2. **Age Group Analysis:** Customers aged 26–35 generated the highest total sales, approximately ₹4.26 crore.

3. **Product Category Analysis:** Food was the top-performing category, generating approximately ₹3.39 crore in sales.

4. **State Analysis:** Uttar Pradesh recorded the highest total sales among the states in the dataset, at approximately ₹1.93 crore.

5. **Occupation Analysis:** Customers in the IT Sector generated the highest total sales by occupation, approximately ₹1.47 crore.

6. **Customer Spending Analysis:** The cleaned dataset contained 3,752 unique customers. Average total spending per customer was approximately ₹28,299.26.

7. **Customer Segmentation:** Of the 3,752 customers, 956 were classified as low spenders, 1,600 as medium spenders, and 1,196 as high spenders, based on the spending thresholds selected for this analysis.
